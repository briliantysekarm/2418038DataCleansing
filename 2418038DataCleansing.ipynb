## Membersihkan Spasi

## Standarisasi Tipe data ada variabel

- standarisasi usia

```python
df["Usia"] = df["Usia"].astype(str).str.extract(r'(\d+)')[0]
df["Usia"] = pd.to_numeric(df["Usia"], errors="coerce")

print(df["Usia"].unique())
```

- standarisasi jenis kelamin

```python
df["Jenis_Kelamin"] = df["Jenis_Kelamin"].str.lower()

df["Jenis_Kelamin"] = df["Jenis_Kelamin"].replace({
    "laki-laki": "Laki-laki",
    "perempuan": "Perempuan"
})

print(df["Jenis_Kelamin"].unique())
```

- Standarisasi Frekuensi Kopi

```python
df["Frekuensi_Kopi"] = (
    df["Frekuensi_Kopi"]
    .astype(str)
    .str.extract(r'(\d+)')[0]
    + " kali/minggu"
)

print(df["Frekuensi_Kopi"].unique())
```

- Standarisasi Jumlah Kopi Harian

```python
df["Jumlah_Kopi_Harian"] = (
    df["Jumlah_Kopi_Harian"]
    .astype(str)
    .str.extract(r'(\d+)')[0]
    + " gelas"
)

print(df["Jumlah_Kopi_Harian"].unique())
```

- Standarisasi Waktu Minum

```python
df["Waktu_Minum"] = df["Waktu_Minum"].str.lower()

df["Waktu_Minum"] = df["Waktu_Minum"].replace({
    "pagi": "Pagi",
    "pagi hari": "Pagi",
    "siang": "Siang",
    "sore": "Sore",
    "malam": "Malam",
    "malam hari": "Malam"
})

print(df["Waktu_Minum"].unique())
```

- Standarisasi Jenis Kopi

```python
df["Jenis_Kopi"] = df["Jenis_Kopi"].str.lower().str.strip()

df["Jenis_Kopi"] = df["Jenis_Kopi"].replace({
    "kopi susu": "Kopi Susu",
    "kopi hitam": "Kopi Hitam",
    "latte": "Latte",
    "cafe latte": "Latte",
    "americano": "Americano",
    "cappuccino": "Cappuccino",
    "tidak minum": "Tidak Minum"
})

print(df["Jenis_Kopi"].unique())
```

- Standarisasi Jam Tidur

```python
def bersihkan_jam(x):
    if pd.isna(x):
        return pd.NA

    x = str(x).lower().strip()
    x = x.replace("wib", "").strip()
    x = x.replace("jam", "").strip()
    x = x.replace(".", ":").strip()
    x = x.replace("pagi", "").strip()

    try:
        return pd.to_datetime(x, format="%H:%M").strftime("%H:%M")
    except:
        return pd.NA

df["Jam_Tidur"] = df["Jam_Tidur"].apply(bersihkan_jam)

print(df["Jam_Tidur"].unique())
```

- Standarisasi Durasi Tidur

```python
df["Durasi_Tidur"] = (
    df["Durasi_Tidur"]
    .astype(str)
    .str.replace(",", ".", regex=False)
    .str.extract(r'(\d+(?:\.\d+)?)')[0]
)

df["Durasi_Tidur"] = pd.to_numeric(
    df["Durasi_Tidur"],
    errors="coerce"
)

print(df["Durasi_Tidur"].unique())
```

- Standarisasi Kualitas Tidur

```python
df["Kualitas_Tidur"] = df["Kualitas_Tidur"].str.lower()

df["Kualitas_Tidur"] = df["Kualitas_Tidur"].replace({
    "baik": "Baik",
    "sedang": "Sedang",
    "buruk": "Buruk"
})

print(df["Kualitas_Tidur"].unique())
```
